```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import os

df = pd.read_csv(
    r"C:\Users\bhumi\Downloads\Datheon\superstore_sales_cleaned.csv"
)

df['Order Date'] = pd.to_datetime(df['Order Date'], errors='coerce')

print("Dataset loaded successfully!")
print("Dataset Shape:", df.shape)

df.head()
```

```python
import os
import matplotlib.pyplot as plt
import seaborn as sns

os.makedirs("visuals", exist_ok=True)

# 1. Monthly Sales
monthly_sales = df.groupby(
    df['Order Date'].dt.to_period('M')
)['Sales'].sum()

plt.figure(figsize=(12, 6))
monthly_sales.plot(kind='line', marker='o')
plt.title("Monthly Sales Trend")
plt.xlabel("Year-Month")
plt.ylabel("Total Sales")
plt.xticks(rotation=45)
plt.tight_layout()
plt.savefig("visuals/monthly_sales.png", dpi=300)
plt.close()


# 2. Category-wise Sales
category_sales = df.groupby('Category')['Sales'].sum()

plt.figure(figsize=(8, 5))
category_sales.plot(kind='bar')
plt.title("Category-wise Sales")
plt.xlabel("Category")
plt.ylabel("Total Sales")
plt.tight_layout()
plt.savefig("visuals/category_sales.png", dpi=300)
plt.close()


# 3. Region-wise Sales
region_sales = df.groupby('Region')['Sales'].sum()

plt.figure(figsize=(8, 5))
region_sales.plot(kind='bar')
plt.title("Region-wise Sales")
plt.xlabel("Region")
plt.ylabel("Total Sales")
plt.tight_layout()
plt.savefig("visuals/region_sales.png", dpi=300)
plt.close()


# 4. Customer Segment Sales
segment_sales = df.groupby('Segment')['Sales'].sum()

plt.figure(figsize=(8, 5))
segment_sales.plot(kind='bar')
plt.title("Customer Segment-wise Sales")
plt.xlabel("Segment")
plt.ylabel("Total Sales")
plt.tight_layout()
plt.savefig("visuals/segment_sales.png", dpi=300)
plt.close()


# 5. Discount vs Profit
plt.figure(figsize=(9, 6))
sns.scatterplot(data=df, x='Discount', y='Profit', alpha=0.5)
plt.title("Discount vs Profit")
plt.tight_layout()
plt.savefig("visuals/discount_profit.png", dpi=300)
plt.close()


print("All 5 visualizations saved successfully!")
```
