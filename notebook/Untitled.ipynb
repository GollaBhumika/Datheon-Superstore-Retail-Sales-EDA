## 4. Data Cleaning and Preprocessing

```python
missing_values = df.isnull().sum()

print("Missing Values:")
print(missing_values)
```

```python
missing_percentage = (df.isnull().sum() / len(df)) * 100

missing_summary = pd.DataFrame({
    "Missing Values": df.isnull().sum(),
    "Missing Percentage": missing_percentage
})

missing_summary
```

```python
print("Duplicate Rows:", df.duplicated().sum())
```

```python
df = df.drop_duplicates()

print("Shape after removing duplicates:", df.shape)
```

```python
# Fill missing numeric values with median
numeric_cols = df.select_dtypes(include=np.number).columns

for col in numeric_cols:
    df[col] = df[col].fillna(df[col].median())

# Fill missing categorical values with mode
categorical_cols = df.select_dtypes(include="object").columns

for col in categorical_cols:
    if df[col].isnull().any() and not df[col].mode().empty:
        df[col] = df[col].fillna(df[col].mode()[0])
```

```python
print("Remaining Missing Values:")
print(df.isnull().sum().sum())

print("Remaining Duplicates:", df.duplicated().sum())
```

## 5. Date Feature Engineering

```python
df.columns.tolist()
```

```python
df["Order Date"] = pd.to_datetime(
    df["Order Date"],
    errors="coerce"
)

df["Year"] = df["Order Date"].dt.year
df["Month"] = df["Order Date"].dt.month
df["Month_Name"] = df["Order Date"].dt.month_name()
df["Year_Month"] = df["Order Date"].dt.to_period("M").astype(str)

df[["Order Date", "Year", "Month", "Month_Name", "Year_Month"]].head()
```

```python
import os

os.makedirs("visuals", exist_ok=True)

print("Visuals folder created successfully!")
```

## 6. Customer Purchase Analysis

```python
customer_sales = (
    df.groupby("Customer Name")["Sales"]
    .sum()
    .sort_values(ascending=False)
)

top_10_customers = customer_sales.head(10)

top_10_customers
```

```python
plt.figure(figsize=(12, 6))

top_10_customers.sort_values().plot(
    kind="barh"
)

plt.title("Top 10 Customers by Sales")
plt.xlabel("Total Sales")
plt.ylabel("Customer Name")
plt.tight_layout()
plt.show()
```

## 7. Monthly Sales Trend Analysis

```python
monthly_sales = (
    df.groupby("Year_Month")["Sales"]
    .sum()
    .sort_index()
)

monthly_sales
```

```python
plt.figure(figsize=(14, 6))

monthly_sales.plot(kind="line", marker="o")

plt.title("Monthly Sales Trend")
plt.xlabel("Year-Month")
plt.ylabel("Total Sales")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

## 8. Category-wise Sales Analysis

```python
category_sales = (
    df.groupby("Category")["Sales"]
    .sum()
    .sort_values(ascending=False)
)

category_sales
```

```python
plt.figure(figsize=(9, 5))

sns.barplot(
    x=category_sales.index,
    y=category_sales.values
)

plt.title("Sales by Product Category")
plt.xlabel("Category")
plt.ylabel("Total Sales")
plt.tight_layout()
plt.show()
```

## 9. Profitability Analysis

```python
category_profit = (
    df.groupby("Category")["Profit"]
    .sum()
    .sort_values(ascending=False)
)

category_profit
```

```python
plt.figure(figsize=(9, 5))

sns.barplot(
    x=category_profit.index,
    y=category_profit.values
)

plt.title("Profit by Product Category")
plt.xlabel("Category")
plt.ylabel("Total Profit")
plt.tight_layout()
plt.show()
```

## 10. Regional Sales Analysis

```python
region_sales = (
    df.groupby("Region")[["Sales", "Profit"]]
    .sum()
    .sort_values(by="Sales", ascending=False)
)

region_sales
```

```python
region_sales["Sales"].plot(
    kind="bar",
    figsize=(9, 5),
    title="Region-wise Sales Performance"
)

plt.xlabel("Region")
plt.ylabel("Total Sales")
plt.tight_layout()
plt.show()
```

## 11. Customer Segment Analysis

```python
segment_sales = (
    df.groupby("Segment")["Sales"]
    .sum()
    .sort_values(ascending=False)
)

segment_sales
```

```python
plt.figure(figsize=(7, 7))

plt.pie(
    segment_sales,
    labels=segment_sales.index,
    autopct="%1.1f%%",
    startangle=90
)

plt.title("Sales Contribution by Customer Segment")
plt.show()
```

## 12. Discount Impact on Sales and Profit

```python
discount_analysis = (
    df.groupby("Discount")[["Sales", "Profit"]]
    .mean()
)

discount_analysis
```

```python
plt.figure(figsize=(10, 6))

sns.scatterplot(
    data=df,
    x="Discount",
    y="Profit",
    hue="Category"
)

plt.title("Discount vs Profit Analysis")
plt.xlabel("Discount")
plt.ylabel("Profit")
plt.tight_layout()
plt.show()
```

## 13. Category-wise Regional Sales Analysis

```python
category_region = pd.pivot_table(
    df,
    values="Sales",
    index="Category",
    columns="Region",
    aggfunc="sum"
)

category_region
```

```python
plt.figure(figsize=(10, 6))

sns.heatmap(
    category_region,
    annot=True,
    fmt=".0f",
    cmap="YlGnBu"
)

plt.title("Category-wise Sales Across Regions")
plt.tight_layout()
plt.show()
```

## 14. Key Business Performance Metrics

```python
total_sales = df["Sales"].sum()
total_profit = df["Profit"].sum()
total_orders = df["Order ID"].nunique()
total_customers = df["Customer ID"].nunique()
average_discount = df["Discount"].mean()

print("Total Sales:", round(total_sales, 2))
print("Total Profit:", round(total_profit, 2))
print("Total Orders:", total_orders)
print("Total Customers:", total_customers)
print("Average Discount:", round(average_discount, 2))
```
